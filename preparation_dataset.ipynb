### 1/ Traitement des données, et création du dataframe et du dataset jason --> réf. 1961-1990

```python
import pandas as pd


# features des données qui seront affichées :
curveFeatures = { curve: featuresAll[curve] for curve in curveToDisplay }
dataSetPanda = {}
dataSetJson = {}

# liste des courbes à tracer directement selon un dataSet 
curveWithDataSet = [curve for curve in curveToDisplay if curveFeatures[curve]["type"] in ["linear", "margin"]]

for curve in curveWithDataSet :

    dataset = pd.read_csv("data/"+sourceFiles[curveFeatures[curve]["source"]])

    if curveFeatures[curve]["type"] == "linear" :
        dataset.drop(["Lower confidence limit (2.5%)", "Upper confidence limit (97.5%)"], axis=1, inplace=True)
        dataset.rename(columns={
            "Time":"date",
            "Anomaly (deg C)":"gap",
        }, inplace=True)
 
    elif curveFeatures[curve]["type"] =="margin":
        dataset.drop(["Anomaly (deg C)"], axis=1, inplace=True)
        dataset.rename(columns={
            "Time":"date",
             "Lower confidence limit (2.5%)":"min",
            "Upper confidence limit (97.5%)":"max",
        }, inplace=True)                

    # ajout du mois quand il n'est pas présent (2023 devient 2023-12, pour correspondre à l'échelle x)
    # Ainsi :
    #   - la moyenne annuelle yyyy est représentée par un point en yyyy-12
    #   - la moyenne annuelle de la dernière année incomplète est quand même en yyyy-12, mais n'est pas tracée.
    if "annual" in curveFeatures[curve]["source"] :        
        dataset["date"] = dataset["date"].apply(lambda d:str(d)+"-12")      #bullet "annuel" au yyyy-12          

    # chargement du dataframe dans un dictionnaire de dataframes :
    dataSetPanda[curve] = dataset
    # chargement du dataframe dans un dictionnaire json :
    dataSetJson[curve] = eval(dataset.to_json(orient="records"))

dataSetPanda['annualconf']
```

### 2/ Traitement des données, et création du panda et du dataset jason --> réf. 1850-1900

##### calcul des moyennes de référence :

```python
def firstmonth(yy, date1):
    # retourne le premier mois de l'année de l'intervalle
    # ex : (2012, 2012-03) retourne 03;  (2014, 2012-03 retourne 01) 
    if yy == int(date1[:4]) and date1[4:]:
         return int(date1[-2:])
    return 1

def lastmonth(yy, date2):
    if yy == int(date2[:4]) and date2[4:]:
         return int(date2[-2:])
    return 12


def temperature_range(curve, date1, date2):
    """retourne la liste des dates comprises entre date1 et date2"""
    if curve=="annual":
#         return [str(yy)+"-01" for yy in range(int(date1[:4]), int(date2[:4])+1)]
        return [str(yy)+"-12" for yy in range(int(date1[:4]), int(date2[:4])+1)]

    elif curve=="monthly":
        return ["{}-{}".format(str(yy), str(mm)) for yy in range(int(date1[:4]), int(date2[:4])+1) for mm in [
            "{}{}".format("0", str(month))[-2:] for month in range(
                firstmonth(yy, date1), 
                lastmonth(yy, date2)+1
            )]           
        ]
    
def temperature_moyenne(curve, date1, date2):
    """calcule la température moyenne de curve entre date1 et date2[inclus]"""
    df = dataSetPanda[curve]
    df_range = df[df['date'].isin(temperature_range(curve, date1, date2))]
    return df_range['gap'].mean()
```

```python
# Calcul des températures moyennes :
daterange = [
    ['1850', '1900'],
    ['1961', '1990'],
    ['1986', '2005'],
    ['2006', '2015'],
    ['2023', '2023'],
]
print("Anomalies de température par rapport à la période 1961-1990 :")
print(f'{"intervalle" : <20} {"moy. mensuelle":20} {"moy.annuelle":15}')

for dr in daterange:
    print(f'{dr[0]+"-"+dr[1] : <20} {temperature_moyenne("monthly", dr[0], dr[1]):10.3f} {temperature_moyenne("annual", dr[0], dr[1]):20.3f}')
print()

print("Anomalies de température par rapport à l'ère pré-industrielle 1850-1900 :")
print(f'{"intervalle" : <20} {"moy. mensuelle":20} {"moy.annuelle":15}')

for dr in daterange:
    dt_mensuel = temperature_moyenne("monthly", dr[0], dr[1]) - temperature_moyenne("monthly", "1850", "1900") 
    dt_annuel = temperature_moyenne("annual", dr[0], dr[1]) - temperature_moyenne("annual", "1850", "1900") 
    print(f'{dr[0]+"-"+dr[1] : <20} { dt_mensuel:10.3f} {dt_annuel:20.3f}')
print()

```

## Pour passer des données HadCRUT5 (réf 1961-1990) aux données IPCC (réf 1850-1900) il faut ajouter +0,36 °C (en fait +0,364°C) 
##### => même si la moyenne 1850-1900 calculée dans le réf HadCRUT5 (réf 1961-1990)  est de -0,356°C parce que dans ce référentiel la moyenne 1961-1990 est de 0,08 °C alors qu'elle devrait être à 0.

IPCC : https://www.ipcc.ch/sr15/chapter/chapter-1/
"In summary, this report adopts a working definition of ‘1.5°C relative to pre-industrial levels’ that corresponds to global average combined land surface air and sea surface temperatures either 1.5°C warmer than the average of the 51-year period 1850–1900, 0.87°C warmer than the 20-year period 1986–2005, or 0.63°C warmer than the decade 2006–2015. These offsets are based on all available published global datasets, combined and updated, which show that 1986–2005 was 0.63°C warmer than 1850–1900 (with a 5–95% range of 0.57°C–0.69°C based on observational uncertainties alone), and 2006–2015 was 0.87°C warmer than 1850–1900 (with a likely range of 0.75°C–0.99°C, also accounting for the possible impact of natural fluctuations). Where possible, estimates of impacts and mitigation pathways are evaluated relative to these more recent periods. Note that the 5–95% intervals often quoted in square brackets in AR5 correspond to very likely ranges, while likely ranges correspond to 17–83%, or the central two-thirds, of the distribution of uncertainty."

##### décalage du dataset pour mettre la référence (0°C) sur la période pré-industrielle 1850-1990 :

```python
dataSetPandaRefPreIndus = dict()
dataSetJsonRefPreIndus = dict()

# décalage mensuel :
month_offset = temperature_moyenne("monthly", "1850", "1900")
annual_offset = temperature_moyenne("annual", "1850", "1900")

for curve in curveWithDataSet :
    dataSetPandaRefPreIndus[curve] = dataSetPanda[curve].copy()


dataSetPandaRefPreIndus["monthly"]["gap"] = dataSetPanda["monthly"]["gap"] - month_offset
dataSetPandaRefPreIndus["monthlyconf"]["min"] = dataSetPanda["monthlyconf"]["min"] - month_offset
dataSetPandaRefPreIndus["monthlyconf"]["max"] = dataSetPanda["monthlyconf"]["max"] - month_offset
dataSetPandaRefPreIndus["annual"]["gap"] = dataSetPanda["annual"]["gap"] - annual_offset
dataSetPandaRefPreIndus["annualconf"]["min"] = dataSetPanda["annualconf"]["min"] - annual_offset
dataSetPandaRefPreIndus["annualconf"]["max"] = dataSetPanda["annualconf"]["max"] - annual_offset


print("Offset mensuel ", month_offset, "Offset annuel: ", annual_offset)
for curve in curveWithDataSet :
    
    dataSetJsonRefPreIndus[curve] = eval(dataSetPandaRefPreIndus[curve].to_json(orient="records"))

    print("Curve / 1961-1990: ", curve)
    display(dataSetPanda[curve])
    print("Curve / 1850-1900: ", curve)
    display(dataSetPandaRefPreIndus[curve])
```
